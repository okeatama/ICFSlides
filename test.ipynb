```python
import requests
from bs4 import BeautifulSoup
from constants import *

url = f"{IMANKATOLIK_BASE_URL}/kalender.php?b=11&t=2025"
# url = "https://imankatolik.or.id/kalender.php?b=11&t=2025"

response = requests.get(url)
soup = BeautifulSoup(response.text, "html.parser")

nov2 = soup.find("a", href="/kalender/2Nov.html")
tr = nov2.find_parent("tr")
readings = tr.next_sibling.next
urls = []

for i in readings.children:
    if i.text == "." or i.text == "<br/>":
        break
    elif i.name != 'a':
        continue

    urls.append(i.attrs["href"])

texts = []
for url in urls:
    new_url = f"{IMANKATOLIK_BASE_URL}{url}"
    response = requests.get(new_url)
    soup = BeautifulSoup(response.text, "html.parser")

    temp = []
    for text in soup.find_all("td", width="95%"):
        temp.append(text.text)
    texts.append("\n".join(temp))

    # texts.append(soup.find("td", width="95%").text)

print(texts[1])
```

```python
import re

def expand_ranges(s):
    # Split by dot or comma
    parts = re.split(r'[.,;]', s)
    numbers = []

    for part in parts:
        if '-' in part:
            start, end = map(int, part.split('-'))
            numbers.extend(range(start, end + 1))
        else:
            numbers.append(int(part))
    return numbers

# Example
s = "1-3.5-6.7.8.10"
result = expand_ranges(s)
print(result)
```
